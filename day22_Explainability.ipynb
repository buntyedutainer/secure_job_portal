```python
# Cell 1: setup and loading the models fixed on Day 20-21 (test set NOT loaded)
import json
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr
from sklearn.base import clone
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_predict
from sklearn.metrics import r2_score, roc_auc_score
from sklearn.tree import DecisionTreeRegressor, export_text
from sklearn.inspection import PartialDependenceDisplay, partial_dependence
from sklearn.ensemble import (RandomForestRegressor, GradientBoostingRegressor,
                              RandomForestClassifier, GradientBoostingClassifier)
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

sns.set_theme(style="whitegrid")
train = pd.read_csv("train.csv")
cfg = json.load(open("final_config.json"))
y = train["G3"]
y_risk = (y < 10).astype(int)
X_all = train.drop(columns="G3")
X_early = X_all.drop(columns=["G1", "G2"])
feature_sets = {"early (no G1/G2)": X_early, "all features": X_all}
tag = {"early (no G1/G2)": "early", "all features": "all"}

def build_reg(name, params):
    base = {"Ridge": make_pipeline(StandardScaler(), Ridge()),
            "Random Forest": RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
            "Gradient Boosting": GradientBoostingRegressor(random_state=42)}[name]
    return base.set_params(**params)

def build_clf(name, params):
    base = {"Logistic": make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000)),
            "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1),
            "Gradient Boosting": GradientBoostingClassifier(random_state=42)}[name]
    return base.set_params(**params)

def load_or_build(path, builder, info, X, target):
    if os.path.exists(path):
        return joblib.load(path)
    print(f"  {path} not found, rebuilding from final_config.json")
    return builder(info["model"], info["params"]).fit(X, target)

reg_models, clf_models = {}, {}
for f, X in feature_sets.items():
    reg_models[f] = load_or_build(f"reg_{tag[f]}_trainonly.joblib", build_reg, cfg["regression"][f], X, y)
    clf_models[f] = load_or_build(f"clf_{tag[f]}_trainonly.joblib", build_clf, cfg["classification"][f], X, y_risk)

kf = KFold(5, shuffle=True, random_state=42)
skf = StratifiedKFold(5, shuffle=True, random_state=42)

print("Models being explained:")
for f in feature_sets:
    r, c = cfg["regression"][f], cfg["classification"][f]
    print(f"  {f:18s} regression: {r['model']:18s} (useful: {r['useful']}) | "
          f"classifier: {c['model']:18s} (usable: {c['usable']})")
print("\nIf 'useful' or 'usable' is False, the explanations below describe a WEAK signal. Say so in your report.")
```

```python
# Cell 2: permutation importance on held-out folds (the engine for most of today)
def cv_perm(model, X, target, cv, predict, metric, groups=None, n_repeats=10, seed=0):
    """Increase in error when a column (or group of columns) is shuffled, per validation fold."""
    rng = np.random.default_rng(seed)
    units = groups if groups is not None else {c: [c] for c in X.columns}
    rows = []
    for tr, va in cv.split(X, target):
        m = clone(model).fit(X.iloc[tr], target.iloc[tr])
        Xv, yv = X.iloc[va], target.iloc[va].values
        base = metric(yv, predict(m, Xv))
        row = {}
        for name, cols in units.items():
            inc = []
            for _ in range(n_repeats):
                Xp = Xv.copy()
                Xp[cols] = Xv[cols].values[rng.permutation(len(Xv))]
                inc.append(metric(yv, predict(m, Xp)) - base)
            row[name] = np.mean(inc)
        rows.append(row)
    return pd.DataFrame(rows)

def summarize(df):
    s = pd.DataFrame({"importance": df.mean(), "std": df.std()})
    s["clearly above zero"] = s["importance"] > 2 * s["std"] / np.sqrt(len(df))
    return s.sort_values("importance", ascending=False)

reg_predict = lambda m, X: m.predict(X)
rmse_metric = lambda a, p: float(np.sqrt(np.mean((a - p) ** 2)))

perm = {f: cv_perm(reg_models[f], X, y, kf, reg_predict, rmse_metric) for f, X in feature_sets.items()}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, f in zip(axes, feature_sets):
    s = summarize(perm[f])
    top = s.head(10)[::-1]
    ax.barh(top.index, top["importance"], xerr=top["std"], color="steelblue")
    ax.set_title(f"Permutation importance, {f}")
    ax.set_xlabel("Increase in RMSE when shuffled (grade points)")
    print(f"\n{f} (top 8):")
    print(s.head(8).round(3).to_string())
    print(f"Features clearly above zero: {int(s['clearly above zero'].sum())} of {len(s)}")
plt.tight_layout()
plt.savefig("perm_importance.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Cell 3: the correlated-features trap, G1 and G2 alone vs together
groups = {c: [c] for c in X_all.columns if c not in ("G1", "G2")}
groups["G1 only"] = ["G1"]
groups["G2 only"] = ["G2"]
groups["G1 + G2 together"] = ["G1", "G2"]

gp = cv_perm(reg_models["all features"], X_all, y, kf, reg_predict, rmse_metric, groups=groups)
print("Increase in RMSE when shuffled (all-features model):")
print(summarize(gp).loc[["G1 only", "G2 only", "G1 + G2 together"], ["importance", "std"]].round(3))
print("\nSum of the two separate values:",
      round(summarize(gp).loc[["G1 only", "G2 only"], "importance"].sum(), 3))
```

```python
# Cell 4: does permutation importance agree with simple correlation?
for f, X in feature_sets.items():
    imp = summarize(perm[f])["importance"]
    cor = X.corrwith(y).abs().reindex(imp.index)
    rho = spearmanr(imp, cor)[0]
    print(f"{f:18s} rank agreement with |correlation with G3|: Spearman {rho:.2f}")
    print(f"{'':18s} top 5 by permutation: {list(imp.index[:5])}")
    print(f"{'':18s} top 5 by correlation: {list(cor.sort_values(ascending=False).index[:5])}\n")
```

```python
# Cell 5: partial dependence and ICE lines for the three most important features
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for i, (f, X) in enumerate(feature_sets.items()):
    top = list(summarize(perm[f]).index[:3])
    PartialDependenceDisplay.from_estimator(
        reg_models[f], X, top, kind="both", subsample=60, random_state=0, ax=axes[i],
        ice_lines_kw={"alpha": 0.15, "color": "steelblue"},
        pd_line_kw={"color": "crimson", "linewidth": 3})
    for ax in axes[i]:
        ax.set_ylim(y.min() - 1, y.max() + 1)
    axes[i][0].set_title(f"{f}: red = average, blue = individual students", loc="left", fontsize=10)
    for feat in top:
        pdres = partial_dependence(reg_models[f], X, [feat], kind="average")
        vals = pdres["grid_values"][0] if "grid_values" in pdres else pdres["values"][0]
        avg = pdres["average"][0]
        print(f"{f:18s} {feat:10s} average prediction at lowest value {avg[0]:.2f}, at highest {avg[-1]:.2f} "
              f"(change {avg[-1] - avg[0]:+.2f}) | feature range {vals[0]:g} to {vals[-1]:g}")
plt.tight_layout()
plt.savefig("pdp_ice.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Cell 6: why did THIS student get THIS prediction? Occlusion against a typical student
def occlusion(model, X, i, ref):
    x = X.iloc[[i]]
    p0 = float(model.predict(x)[0])
    contrib = {}
    for c in X.columns:
        xr = x.copy()
        xr[c] = ref[c]
        contrib[c] = p0 - float(model.predict(xr)[0])
    return p0, pd.Series(contrib)

def pick_students(p):
    order = np.argsort(p)
    at_risk_idx = np.where(y.values < 10)[0]
    return {"lowest predicted": order[0],
            "typical": order[len(order) // 2],
            "highest predicted": order[-1],
            "at risk, predicted high": at_risk_idx[np.argmax(p[at_risk_idx])]}

fig, axes = plt.subplots(2, 4, figsize=(20, 8))
for r, (f, X) in enumerate(feature_sets.items()):
    model = reg_models[f]
    ref = X.median()                                           # the "typical student"
    base_pred = float(model.predict(pd.DataFrame([ref]))[0])
    oof = cross_val_predict(clone(model), X, y, cv=kf)         # used only to choose which students to show
    for c, (label, idx) in enumerate(pick_students(oof).items()):
        p0, contrib = occlusion(model, X, idx, ref)
        top = contrib.reindex(contrib.abs().sort_values(ascending=False).index).head(6)[::-1]
        axes[r, c].barh(top.index, top.values, color=np.where(top.values > 0, "seagreen", "indianred"))
        axes[r, c].axvline(0, color="black", linewidth=0.8)
        axes[r, c].set_title(f"{f}\n{label}\nactual {y.iloc[idx]}, predicted {p0:.1f} (typical student: {base_pred:.1f})",
                             fontsize=9)
        gap = p0 - base_pred
        print(f"{f:18s} {label:26s} prediction minus typical = {gap:+.2f} | sum of contributions = {contrib.sum():+.2f}")
plt.tight_layout()
plt.savefig("local_explanations.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Cell 7: "what if" questions, and why the answers are NOT what a school could achieve
scenarios = [
    ("failures set to 0 (students who have failures)", "failures", lambda s: 0, lambda X: X["failures"] > 0),
    ("wants higher education set to yes", "higher", lambda s: 1, lambda X: X["higher"] == 0),
    ("study time up one level", "studytime", lambda s: np.minimum(s + 1, 4), lambda X: X["studytime"] < 4),
    ("absences set to 0", "absences", lambda s: 0, lambda X: X["absences"] > 0),
]
raw = {"failures": y[X_all["failures"] == 0].mean() - y[X_all["failures"] > 0].mean(),
       "higher": y[X_all["higher"] == 1].mean() - y[X_all["higher"] == 0].mean()}

rows = []
for f, X in feature_sets.items():
    for label, col, new_val, mask_fn in scenarios:
        mask = mask_fn(X)
        X2 = X.copy()
        X2.loc[mask, col] = new_val(X.loc[mask, col])
        delta = reg_models[f].predict(X2[mask]) - reg_models[f].predict(X[mask])
        rows.append({"features": f, "what-if": label, "students affected": int(mask.sum()),
                     "mean change in predicted G3": round(delta.mean(), 2),
                     "raw group gap in data": round(raw[col], 2) if col in raw else ""})
pd.DataFrame(rows)
```

```python
# Cell 8: a global surrogate, a small readable tree that imitates each model
for f, X in feature_sets.items():
    p = reg_models[f].predict(X)
    sur = DecisionTreeRegressor(max_depth=3, min_samples_leaf=10, random_state=0)
    fidelity = r2_score(p, cross_val_predict(sur, X, p, cv=kf))
    sur.fit(X, p)
    print(f"\n=== {f}: surrogate tree, fidelity R2 = {fidelity:.2f} (1.0 means a perfect imitation) ===")
    print(export_text(sur, feature_names=list(X.columns), decimals=1))
```

```python
# Cell 9: how stable are the rankings? (the five folds are five slightly different samples)
for f in feature_sets:
    d = perm[f]
    rank_cors = [spearmanr(d.iloc[i], d.iloc[j])[0] for i in range(len(d)) for j in range(i + 1, len(d))]
    tops = [set(d.iloc[i].sort_values(ascending=False).index[:3]) for i in range(len(d))]
    overlap = np.mean([len(tops[i] & tops[j]) / 3 for i in range(len(d)) for j in range(i + 1, len(d))])
    always = set.intersection(*tops)
    print(f"{f:18s} average rank agreement between folds (Spearman): {np.mean(rank_cors):.2f}"
          f" | average top-3 overlap: {overlap:.0%} | in the top 3 of every fold: {sorted(always) or 'nothing'}")
```

```python
# Cell 10: explaining the at-risk classifier (importance = drop in ROC AUC when shuffled)
clf_predict = lambda m, X: m.predict_proba(X)[:, 1]
auc_drop = lambda a, p: -roc_auc_score(a, p)

permc = {f: cv_perm(clf_models[f], X, y_risk, skf, clf_predict, auc_drop, n_repeats=8)
         for f, X in feature_sets.items()}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, f in zip(axes, feature_sets):
    s = summarize(permc[f])
    top = s.head(10)[::-1]
    ax.barh(top.index, top["importance"], xerr=top["std"], color="darkorange")
    ax.set_title(f"At-risk classifier, {f}")
    ax.set_xlabel("Drop in ROC AUC when shuffled")
    print(f"\n{f} (top 6):")
    print(s.head(6).round(3).to_string())
    print(f"Features clearly above zero: {int(s['clearly above zero'].sum())} of {len(s)}")
plt.tight_layout()
plt.savefig("perm_importance_classifier.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Cell 11: save the results for the report
out = []
for task, d in [("regression: RMSE increase", perm), ("at-risk classifier: AUC drop", permc)]:
    for f in feature_sets:
        s = summarize(d[f]).reset_index().rename(columns={"index": "feature"})
        s["features"], s["task"] = f, task
        out.append(s)
pd.concat(out).to_csv("explainability_importance.csv", index=False)
print("Saved explainability_importance.csv, perm_importance.png, pdp_ice.png, local_explanations.png, "
      "perm_importance_classifier.png")
```
